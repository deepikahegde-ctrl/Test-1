## Exercise-1

```python
num = int(input('Enter the number')) #taking a number from the user

if num <= 1:
    print(num,'The number is not a prime number') #if the number is less than or equal to 1 that will not be a prime number
else:
    for i in range(2,num): # taking the number from 2 to num-1
        if num % i == 0: #checking the number that the user is entered is divisible by its lower number or not
            print(num,'The number is not a prime number')
            break; # not to move further
    else:
        print(num,'The number is a prime number')
            
```

## Exercise-2

```python
import random

num1 = random.randint(1,100) ## Generating 2 random numbers using random library
num2 = random.randint(1,100)

product=int(input(f"What is the product of {num1} and {num2}?")) #asking the user to guess the number
corr_prod=num1*num2
if product == corr_prod: #checking the user answer and the corr_prod
    print("You are correct, Good!")
else:
    print(f"No, The correct answer is {corr_prod}")
```

```python
room_charge = 2000*4
doctor_cons = 1500
lab_test = 300*3
med_charge = 2400 # given variables
total = room_charge + doctor_cons + lab_test + med_charge # total bill
if total > 10000: #checking if the total is exceeds 10000 or not
    discount = total * 10 / 100 # calculating the discount
else:
    discount = 0 
print("total bill is",total) #printing the total bill
print("total discount is",discount) #printing the discount
```
