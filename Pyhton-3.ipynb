## Descriptive Analytics for Numerical Columns

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("sales_data_with_discounts.csv") #loading the csv data set into the jupyter notebook
```

```python
df
```

```python
df.info()
```

```python
numerical_col = df.select_dtypes(include="number").columns #checking that how many numerical columns are there in the data set
numerical_col
```

```python
df.describe() #we can calculate the mean, median, mode and standard deviation using describe() function and can also find as below
```

```python
df.Volume.mean() #calculating the mean of the Volume column using mean()
```

```python
df.Volume.median() #calculating the median of the column using median() 
```

```python
df.Volume.mode() #mode can be multiple so that its giving the index number also
```

```python
df.Volume.std() #calculating the standard deviation using the function std()
```

## Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
for i in range(len(numerical_col)):
    print(i)
```

```python
plt.figure(figsize=(12,8))
for i in range(len(numerical_col)):
    plt.subplot(3,3,i+1)
    sns.histplot(data=df[numerical_col[i]])
    plt.tight_layout()
```

```python
plt.figure(figsize=(12,8))
for i in range(len(numerical_col)):
    plt.subplot(3,3,i+1)
    sns.boxplot(data=df[numerical_col[i]])
    plt.tight_layout()
```

## Bar Chart Analysis for Categorical Column:

```python
cat_col = df.select_dtypes(include='object').columns
cat_col
```

```python
plt.figure(figsize=(20,8))
for i in range(len(cat_col)):
    plt.subplot(3,3,i+1)
    sns.countplot(x=df[cat_col[i]])
    plt.tight_layout()
```
