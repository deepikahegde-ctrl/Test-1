### Estimation and Confidence Interval

```python
import numpy as np
import statsmodels
import scipy.stats as stats
import warnings
warnings.filterwarnings('ignore')
```

```python
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 
        1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
data =np.array(data) # converting it into numpy array
data
```

### Part-A

```python
n = len(data) # sample size
x_bar = np.mean(data) #sample mean
s = np.std(data,ddof=1) #standard deviation of the sample data
df=n-1  #degree of freedom
alpha = 0.01
```

```python
# we have to go with the t-test or t-distribution because we do not know the value of the std_dev of population and the sample size is less than 30
#tcritical value
t_critic =  stats.t.ppf(1-alpha/2,df)
t_critic
```

```python
# standard error
SE = s/np.sqrt(n)
SE
```

```python
# margine of error
ME_t = t_critic * SE
ME_t
```

```python
# Confidence Interval
CI = (x_bar - ME_t, x_bar + ME_t)
CI
```

### Part-B

```python
# known populatin is 0.2 millian characters so we can use z_test or z-distribution
sigma = 0.2
z_crit = stats.norm.ppf(1-alpha/2)
z_crit
```

```python
# standard error
SE_z = sigma/np.sqrt(n)
SE_z
```

```python
# margine error
ME_z = z_crit + SE_z
ME_z
```

```python
# Confidence Interval
CI = (x_bar - ME_z, x_bar + ME_z)
CI
```
