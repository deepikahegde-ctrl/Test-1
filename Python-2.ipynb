## Section C: Exploring Pandas

```python
data = {'Name':['Shilpa','Mythri','Shwetha','Shruthi','Divya','Uday','Akshay','Vinay','Chinamay','Divya'],
    'Age':[23,21,33,31,32,33,34,35,25,32],
    'City':['Bangalore','Pune','Kodagu','Kashmir','Bangalore','Bangalore','Chennai','Vaynad','Hyderabad','Bangalore'],
    'Salary':[50000,35000,42000,60000,45000,35000,None,40000,None,45000,]
}
```

```python
df3 = pd.DataFrame(data) 
df3
```

```python
df3.info()
```

```python
df3.describe()
```

```python
print(df3[['Name','City']]) #selecting only the name and city column
```

```python
df3.drop('City',axis=1,inplace=True) #here axis=1 means the column and inplace=True is we deleted the city column permenantly in df3.
                                     #df3 itself is changed permenantly
df3
```

```python
df3['Salary'].fillna(df3['Salary'].mean()) #taking the salary columns and filling the Nan value that is missing values with the mean value.
```

```python
df3.drop_duplicates(inplace=True)  #dropping the duplicate value in df3 permenantly
df3
```
